```python
l_column = list(df.columns) # Making a list out of column names
len_feature = len(l_column) # Length of column vector list
l_column
```

```python
X = df[l_column[0:len_feature-2]]
y = df[l_column[len_feature-2]]
```

```python
print("Feature set size:",X.shape)
print("Variable set size:",y.shape)
```

```python
X.head()
```

```python
y.head()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y,
                            test_size=0.3, random_state=123)
```

```python
from sklearn.linear_model import LinearRegression
lm = LinearRegression() # Creating a Linear Regression object 'lm'
```

```python
X_train_numeric = X_train.drop('animal_name', axis=1)
lm.fit(X_train_numeric,y_train)
```

```python
print("The intercept term of the linear model:", lm.intercept_)
```

```python
print("The coefficients of the linear model:", lm.coef_)
```

```python
import pandas as pd

# Create DataFrame of coefficients
cdf = pd.DataFrame(data=lm.coef_, index=X_train_numeric.columns, columns=['Coefficient'])

# Include intercept
intercept_df = pd.DataFrame(data=[lm.intercept_], index=['Intercept'], columns=['Coefficient'])

# Concatenate intercept and coefficients
cdf = pd.concat([intercept_df, cdf])
cdf
```

```python
import numpy as np
import pandas as pd

# n = number of training samples
n = X_train_numeric.shape[0]

# k = number of features used for fitting
k = X_train_numeric.shape[1]

# degrees of freedom
dfN = n - k

# Predicted values on training set
train_pred = lm.predict(X_train_numeric)

# Squared errors
train_error = np.square(train_pred - y_train)

# Sum of squared errors
sum_error = np.sum(train_error)

# Initialize list for standard errors
se = []

# Compute standard error for each feature
for feature in X_train_numeric.columns:
    r = (sum_error / dfN)
    r = r / np.sum(np.square(X_train_numeric[feature] - X_train_numeric[feature].mean()))
    se.append(np.sqrt(r))

# Add np.nan for the intercept's standard error to match cdf length
se.insert(0, np.nan)

# Add Standard Error and t-statistic to the coefficients DataFrame
cdf['Standard Error'] = se
cdf['t-statistic'] = cdf['Coefficient'] / cdf['Standard Error']

cdf
```

```python
print("Therefore, features arranged in the order of importance for predicting the animal type\n", '-'*90, sep='')

# Sort features by absolute t-statistic (higher magnitude = more important)
l = list(cdf.sort_values('t-statistic', ascending=False).index)

# Print in readable format
print(' > \n'.join(l))
```

```python
import matplotlib.pyplot as plt
from matplotlib import gridspec
import seaborn as sns

# Map type to readable categories
type_map = {
    1: "Mammal",
    2: "Bird",
    3: "Reptile",
    4: "Fish",
    5: "Amphibian",
    6: "Insect",
    7: "Invertebrate"
}

df["category"] = df['type'].map(type_map)

# Get top 5 features from cdf (ordered by t-statistic), excluding 'Intercept'
l = list(cdf.sort_values('t-statistic', ascending=False).index[1:6])

fig = plt.figure(figsize=(18, 10))
gs = gridspec.GridSpec(2, 3)

for i in range(len(l)):
    ax = plt.subplot(gs[i])
    # Use stripplot to handle categorical target
    sns.stripplot(x=df[l[i]], y=df['category'], ax=ax, jitter=True)
    ax.set_title(f"{l[i]} vs. Category", fontdict={'fontsize': 20})
    ax.set_xlabel(l[i])
    ax.set_ylabel('Category')

plt.tight_layout()
plt.show()
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Example:
train_pred = lm.predict(X_train_numeric) # Use X_train_numeric, which was used for fitting

# Note: For accuracy_score and classification_report, y_train (the true labels)
# should ideally be compared with classification predictions (0 or 1), not raw regression outputs.
# For regression, metrics like R2 score, MAE, MSE are more appropriate.
# If 'catsize' is intended as a categorical outcome, consider a classifier model instead of LinearRegression.
# For now, let's cast regression predictions to binary for these metrics.
# You might need to set a threshold for converting 'train_pred' to binary classes if 'catsize' is 0/1.
# Assuming 'catsize' is binary (0 or 1) and a threshold of 0.5 for prediction
train_pred_binary = (train_pred >= 0.5).astype(int)

print("Accuracy:", round(accuracy_score(y_train, train_pred_binary), 3))
print("Classification Report:\n", classification_report(y_train, train_pred_binary))
```

```python
predictions = lm.predict(X_test.drop('animal_name', axis=1))
print ("Type of the predicted object:", type(predictions))
print ("Size of the predicted object:", predictions.shape)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

# Convert continuous predictions to binary (0 or 1) using a threshold
predictions_binary = (predictions >= 0.5).astype(int)

# Assuming predictions = model.predict(X_test)
cm = confusion_matrix(y_test, predictions_binary)

plt.figure(figsize=(10,7))
plt.title("Actual vs Predicted Animal Types", fontsize=25)

# Display confusion matrix with correct binary labels
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=[0, 1])
disp.plot(cmap=plt.cm.Blues, ax=plt.gca())
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Compute "residuals" as numeric difference (predicted class - actual class)
residuals = predictions - y_test

plt.figure(figsize=(10,7))
plt.title("Histogram of prediction errors", fontsize=25)
plt.xlabel("Prediction error (Predicted - Actual)", fontsize=18)
plt.ylabel("Count", fontsize=18)

sns.histplot(residuals, kde=True, bins=7)  # bins=number of classes
plt.show()
```

```python
import matplotlib.pyplot as plt

# Compute prediction errors as numeric differences
errors = predictions - y_test

plt.figure(figsize=(10,7))
plt.title("Prediction errors vs Predicted Classes", fontsize=25)
plt.xlabel("Predicted Animal Type (numeric)", fontsize=18)
plt.ylabel("Prediction Error (Predicted - Actual)", fontsize=18)

plt.scatter(predictions, errors, alpha=0.7)
plt.axhline(0, color='red', linestyle='--')  # reference line for perfect predictions
plt.show()
```
