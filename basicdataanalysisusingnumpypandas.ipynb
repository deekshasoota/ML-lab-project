```python
import matplotlib.pyplot as plt
```

```python
plt.plot(df["legs"])
plt.title("Line Plot of Number of Legs")
plt.xlabel("Index")
plt.ylabel("Number of Legs")
plt.show()
```

```python
df["type"].value_counts().plot(kind="bar")
plt.title("Bar Chart of Animal Type Count")
plt.xlabel("Animal Type")
plt.ylabel("Count")
plt.show()
```

```python
plt.scatter(df["legs"], df["milk"])
plt.title("Scatter Plot of Legs vs Milk")
plt.xlabel("Number of Legs")
plt.ylabel("Milk (0 = No, 1 = Yes)")
plt.show()
```

```python
plt.hist(df["legs"])
plt.title("Histogram of Number of Legs")
plt.xlabel("Number of Legs")
plt.ylabel("Frequency")
plt.show()
```

```python
q1 = df["legs"].quantile(0.25)
q2 = df["legs"].quantile(0.50)
q3 = df["legs"].quantile(0.75)
iqr = q3 - q1

# Box plot
plt.boxplot(df["legs"])
plt.title("Box Plot of Number of Legs")
plt.ylabel("Number of Legs")

# Label quartiles and IQR
plt.text(1.05, q1, f"Q1 = {q1:.2f}", va="top")
plt.text(1.05, q2, f"Q2 = {q2:.2f}", va="center")
plt.text(1.05, q3, f"Q3 = {q3:.2f}", va="bottom")
plt.text(1.05, (q1 + q3)/2, f"IQR = {iqr:.2f}", va="center", fontweight="bold")

plt.show()
```

```python
# Select numeric columns from Zoo dataset
data = df[["legs", "milk", "eggs", "airborne", "aquatic"]]

plt.figure(figsize=(20, 5))

# Create box plot
box = plt.boxplot(
    data.values,
    tick_labels=data.columns,
    patch_artist=True,
    medianprops=dict(color="red", linewidth=2),
    boxprops=dict(facecolor="lightblue"),
    whiskerprops=dict(color="black"),
    capprops=dict(color="black")
)

# Labels and title
plt.xlabel("Zoo Features")
plt.ylabel("Value")
plt.title("Box Plot of Zoo Dataset Features with Quartiles")

# Calculate quartiles
Q1 = data.quantile(0.25)
Q2 = data.quantile(0.50)
Q3 = data.quantile(0.75)

# Annotate Q1, Median, Q3
for i, col in enumerate(data.columns, start=1):
    plt.text(i, Q1[col] - 0.1, f"Q1 = {Q1[col]:.2f}", fontsize=9, color="blue", ha="center")
    plt.text(i, Q2[col], f"Median = {Q2[col]:.2f}", fontsize=9, color="red", ha="center", va="bottom")
    plt.text(i, Q3[col] + 0.1, f"Q3 = {Q3[col]:.2f}", fontsize=9, color="green", ha="center")

# Optional: grid
plt.grid(axis="y", linestyle="--", alpha=0.7)

plt.show()
```
